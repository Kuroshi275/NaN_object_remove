Імпортуємо необхідну бібліотеку

```python
import pandas as pd
```

Зчитуємо та виводимо датасет

```python
ds = pd.read_csv("https://raw.githubusercontent.com/Kuroshi275/NaN_object_remove/refs/heads/main/video%20games%20sales.csv")
pd.set_option('display.expand_frame_repr', False)

print(ds)
```

Перевіряємо у яких совпцях присутнє **"NaN"**

```python
print(ds.isnull().any())
```

Виводимо усі рядки з **"NaN"** та кількість **"NaN"** у кожному стовпцю

```python
print(ds[ds[["Year", "Publisher"]].isnull().any(axis=1)], "\n")
print("NaN count: \n", ds[["Year", "Publisher"]].isnull().sum())
```

Заміняємо **"NaN"** на **"0.0"**, також робимо перевірку чи все замінилось

```python
ds_replace = ds.fillna(0.0)
print(ds_replace, "\n")
print("NaN count: \n", ds_replace[["Year", "Publisher"]].isnull().sum())
```

Створюємо стовпець з сумою продаж з Америки, Японії та Європи

```python
sales_cols = ["NA_Sales", "EU_Sales", "JP_Sales"]
ds["Main_Sales"] = ds[sales_cols].sum(axis=1)
print(ds)
```

Видаляємо усі рядки з **"NaN"** та виводимо зміну у кількості рядків

```python
ds_purify = ds.dropna(subset=["Year", "Publisher"])
print(ds_purify)

print("Rows before cleaning:", ds.shape[0])
print("Rows after cleaning:", ds_purify.shape[0])
print("Rows dif:", ds.shape[0] - ds_purify.shape[0])
```

Виводимо топ продаваємих ігор за весь час до 2020 року

```python
top10_games = ds.nlargest(10, "Global_Sales")[["Name", "Global_Sales"]]
print(top10_games)
```

Виводимо повну інформацію про топ продаваємих ігор у 2016 році

```python
top10_2016 = ds[ds["Year"] == 2016].nlargest(10, "Global_Sales")
print(top10_2016)
```
